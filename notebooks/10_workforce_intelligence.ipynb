```python
# ============================================================
# IMPORT LIBRARIES
# ============================================================

import numpy as np
import pandas as pd

from pathlib import Path

print("✓ Libraries imported successfully.")
```

```python
# ============================================================
# PROJECT PATHS
# ============================================================

CURRENT_DIR = Path.cwd()
PROJECT_ROOT = CURRENT_DIR.parent

RAW_PATH = PROJECT_ROOT / "data" / "raw"
INTEGRATED_PATH = PROJECT_ROOT / "data" / "integrated"
OUTPUT_PATH = PROJECT_ROOT / "data" / "eda_outputs"

OUTPUT_PATH.mkdir(
    parents=True,
    exist_ok=True
)

print("Project Root:")
print(PROJECT_ROOT)

print("\nRaw Data:")
print(RAW_PATH)

print("\nIntegrated Data:")
print(INTEGRATED_PATH)

print("\nOutput Data:")
print(OUTPUT_PATH)
```

```python
# ============================================================
# LOCATE ORIGINAL DATASETS
# ============================================================

def find_file(filename):

    candidates = [
        RAW_PATH / filename,
        PROJECT_ROOT / "data" / "processed" / filename,
        PROJECT_ROOT / "data" / "integrated" / filename
    ]

    for path in candidates:
        if path.exists():
            return path

    raise FileNotFoundError(
        f"Could not find {filename}"
    )


# EXACT ORIGINAL DATASET FILENAMES

attrition_file = find_file(
    "employee_attrition.csv"
)

performance_file = find_file(
    "hr_performance_engagement.csv"
)

occupation_file = find_file(
    "occupation_data.csv"
)

essential_file = find_file(
    "essential_skills.csv"
)

software_file = find_file(
    "software_skills.csv"
)


print("✓ All 5 original datasets located.")

print("\nFiles:")
print("✓", attrition_file.name)
print("✓", performance_file.name)
print("✓", occupation_file.name)
print("✓", essential_file.name)
print("✓", software_file.name)
```

```python
# ============================================================
# LOAD ORIGINAL DATASETS
# ============================================================

attrition_df = pd.read_csv(attrition_file)
performance_df = pd.read_csv(performance_file)
occupation_df = pd.read_csv(occupation_file)
essential_df = pd.read_csv(essential_file)
software_df = pd.read_csv(software_file)

print("DATASET SHAPES")
print("=" * 60)

print(f"Employee Attrition       : {attrition_df.shape}")
print(f"Performance & Engagement : {performance_df.shape}")
print(f"Occupation Data          : {occupation_df.shape}")
print(f"Essential Skills         : {essential_df.shape}")
print(f"Software Skills          : {software_df.shape}")
```

```python
# ============================================================
# PERFORMANCE & ENGAGEMENT DATA
# ============================================================

print("Performance & Engagement Columns:")
print("=" * 60)

for col in performance_df.columns:
    print(f"✓ {col}")
```

```python
# ============================================================
# DEPARTMENT ENGAGEMENT INTELLIGENCE
# ============================================================

department_engagement = (
    performance_df
    .groupby("Department")
    .agg(
        Employee_Count=("Employee ID", "nunique"),
        Avg_Performance_Score=("Performance Score", "mean"),
        Avg_KPI_Score=("KPI Score", "mean"),
        Avg_Attendance=("Attendance (%)", "mean"),
        Avg_Peer_Rating=("Peer Rating", "mean"),
        Avg_Task_Completion=("Task Completion (%)", "mean"),
        Avg_Work_Hours=("Work Hours Logged", "mean"),
        Avg_Manager_Feedback=("Manager Feedback", "mean"),
        Avg_Training_Hours=("Training Hours", "mean")
    )
    .reset_index()
)

print("✓ Department engagement metrics calculated.")

display(department_engagement)
```

```python
# ============================================================
# ENGAGEMENT INDEX
# ============================================================

engagement_components = [
    "Avg_Performance_Score",
    "Avg_KPI_Score",
    "Avg_Attendance",
    "Avg_Peer_Rating",
    "Avg_Task_Completion",
    "Avg_Manager_Feedback"
]

for col in engagement_components:
    min_value = department_engagement[col].min()
    max_value = department_engagement[col].max()

    if max_value == min_value:
        department_engagement[col + "_Norm"] = 1.0
    else:
        department_engagement[col + "_Norm"] = (
            (department_engagement[col] - min_value)
            / (max_value - min_value)
        )

department_engagement["Engagement_Index"] = (
    department_engagement[
        [col + "_Norm" for col in engagement_components]
    ]
    .mean(axis=1)
    * 100
)

print("✓ Engagement Index calculated on a 0–100 scale.")

display(
    department_engagement[
        [
            "Department",
            "Employee_Count",
            "Engagement_Index"
        ]
    ]
)
```

```python
# ============================================================
# ENGAGEMENT LEVEL
# ============================================================

def engagement_level(score):

    if score >= 70:
        return "HIGH"

    elif score >= 40:
        return "MEDIUM"

    else:
        return "LOW"


department_engagement["Engagement_Level"] = (
    department_engagement["Engagement_Index"]
    .apply(engagement_level)
)

print("✓ Engagement levels assigned.")

display(
    department_engagement[
        [
            "Department",
            "Engagement_Index",
            "Engagement_Level"
        ]
    ]
)
```

```python
# ============================================================
# JOB ROLE INTELLIGENCE
# ============================================================

role_intelligence = (
    performance_df
    .groupby(["Department", "Job Role"])
    .agg(
        Employee_Count=("Employee ID", "nunique"),
        Avg_Performance_Score=("Performance Score", "mean"),
        Avg_KPI_Score=("KPI Score", "mean"),
        Avg_Attendance=("Attendance (%)", "mean"),
        Avg_Task_Completion=("Task Completion (%)", "mean"),
        Avg_Training_Hours=("Training Hours", "mean"),
        Promotion_Eligible_Count=(
            "Promotion Eligibility",
            lambda x: (x == "Yes").sum()
        )
    )
    .reset_index()
)

role_intelligence["Promotion_Eligibility_Rate"] = (
    role_intelligence["Promotion_Eligible_Count"]
    / role_intelligence["Employee_Count"]
    * 100
)

print("✓ Job role intelligence calculated.")

display(role_intelligence.head(20))
```

```python
# ============================================================
# O*NET OCCUPATION INTELLIGENCE
# ============================================================

occupation_intelligence = occupation_df[
    [
        "O*NET-SOC Code",
        "Title",
        "Description"
    ]
].copy()

print(
    f"✓ Occupation records: "
    f"{len(occupation_intelligence)}"
)

display(
    occupation_intelligence.head(10)
)
```

```python
# ============================================================
# ESSENTIAL SKILLS PREPARATION
# ============================================================

essential_skills = essential_df.copy()

essential_skills["Data Value"] = pd.to_numeric(
    essential_skills["Data Value"],
    errors="coerce"
)

essential_skills = essential_skills.dropna(
    subset=[
        "O*NET-SOC Code",
        "Title",
        "Element Name",
        "Scale Name",
        "Data Value"
    ]
)

print(
    f"✓ Valid essential-skill records: "
    f"{len(essential_skills)}"
)

print("\nScale types found:")
print(
    essential_skills["Scale Name"].value_counts()
)
```

```python
# ============================================================
# ESSENTIAL SKILLS PREPARATION
# ============================================================

essential_skills = essential_df.copy()

essential_skills["Data Value"] = pd.to_numeric(
    essential_skills["Data Value"],
    errors="coerce"
)

essential_skills = essential_skills.dropna(
    subset=[
        "O*NET-SOC Code",
        "Title",
        "Element Name",
        "Scale Name",
        "Data Value"
    ]
)

print(
    f"✓ Valid essential-skill records: "
    f"{len(essential_skills)}"
)

print("\nScale types found:")
print(
    essential_skills["Scale Name"].value_counts()
)
```

```python
# ============================================================
# SOFTWARE SKILLS PREPARATION
# ============================================================

software_skills = software_df.copy()

software_skills = software_skills.dropna(
    subset=[
        "O*NET-SOC Code",
        "Title",
        "Element Name"
    ]
)

print(
    f"✓ Valid software-skill records: "
    f"{len(software_skills)}"
)

print("\nIn Demand values:")
print(
    software_skills["In Demand"].value_counts(dropna=False)
)

print("\nHot Technology values:")
print(
    software_skills["Hot Technology"].value_counts(dropna=False)
)
```

```python
# ============================================================
# SOFTWARE SKILL INTELLIGENCE
# ============================================================

software_skill_requirements = (
    software_skills
    .groupby(
        [
            "O*NET-SOC Code",
            "Title",
            "Element Name"
        ],
        as_index=False
    )
    .agg(
        Hot_Technology=("Hot Technology", "max"),
        In_Demand=("In Demand", "max"),
        Workplace_Example=("Workplace Example", "first")
    )
)

print(
    f"✓ Software skill requirement records: "
    f"{len(software_skill_requirements)}"
)

display(
    software_skill_requirements.head(10)
)
```

```python
# ============================================================
# RECOVER ESSENTIAL SKILL REQUIREMENTS
# ============================================================

essential_skills = essential_df.copy()

essential_skills["Data Value"] = pd.to_numeric(
    essential_skills["Data Value"],
    errors="coerce"
)

essential_skills = essential_skills.dropna(
    subset=[
        "O*NET-SOC Code",
        "Title",
        "Element Name",
        "Scale Name",
        "Data Value"
    ]
)

essential_skill_requirements = (
    essential_skills
    .pivot_table(
        index=[
            "O*NET-SOC Code",
            "Title",
            "Element Name"
        ],
        columns="Scale Name",
        values="Data Value",
        aggfunc="mean"
    )
    .reset_index()
)

essential_skill_requirements.columns.name = None

essential_skill_requirements = (
    essential_skill_requirements
    .rename(
        columns={
            "Importance": "Skill_Importance",
            "Level": "Skill_Level"
        }
    )
)

print("✓ Essential skill requirements recreated.")
print(f"Records: {len(essential_skill_requirements)}")

display(essential_skill_requirements.head(10))
```

```python
# ============================================================
# RECOVER SOFTWARE SKILL REQUIREMENTS
# ============================================================

software_skills = software_df.copy()

software_skills = software_skills.dropna(
    subset=[
        "O*NET-SOC Code",
        "Title",
        "Element Name"
    ]
)

software_skill_requirements = (
    software_skills
    .groupby(
        [
            "O*NET-SOC Code",
            "Title",
            "Element Name"
        ],
        as_index=False
    )
    .agg(
        Hot_Technology=("Hot Technology", "max"),
        In_Demand=("In Demand", "max"),
        Workplace_Example=("Workplace Example", "first")
    )
)

print("✓ Software skill requirements recreated.")
print(f"Records: {len(software_skill_requirements)}")

display(software_skill_requirements.head(10))
```

```python
# ============================================================
# COMBINED ROLE-SKILL INTELLIGENCE
# ============================================================

essential_combined = (
    essential_skill_requirements[
        [
            "O*NET-SOC Code",
            "Title",
            "Element Name",
            "Skill_Importance",
            "Skill_Level"
        ]
    ]
    .copy()
)

essential_combined["Skill_Type"] = "Essential"
essential_combined["Hot_Technology"] = np.nan
essential_combined["In_Demand"] = np.nan
essential_combined["Workplace_Example"] = np.nan


software_combined = (
    software_skill_requirements[
        [
            "O*NET-SOC Code",
            "Title",
            "Element Name",
            "Hot_Technology",
            "In_Demand",
            "Workplace_Example"
        ]
    ]
    .copy()
)

software_combined["Skill_Type"] = "Software"
software_combined["Skill_Importance"] = np.nan
software_combined["Skill_Level"] = np.nan


role_skill_requirements = pd.concat(
    [
        essential_combined,
        software_combined[
            [
                "O*NET-SOC Code",
                "Title",
                "Element Name",
                "Skill_Importance",
                "Skill_Level",
                "Skill_Type",
                "Hot_Technology",
                "In_Demand",
                "Workplace_Example"
            ]
        ]
    ],
    ignore_index=True
)

print(
    f"✓ Combined role-skill records: "
    f"{len(role_skill_requirements)}"
)

display(
    role_skill_requirements.head(10)
)
```

```python
# ============================================================
# ROLE SKILL SUMMARY
# ============================================================

role_skill_summary = (
    role_skill_requirements
    .groupby(
        [
            "O*NET-SOC Code",
            "Title"
        ]
    )
    .agg(
        Total_Skills=("Element Name", "nunique"),
        Essential_Skills=(
            "Skill_Type",
            lambda x: (x == "Essential").sum()
        ),
        Software_Skills=(
            "Skill_Type",
            lambda x: (x == "Software").sum()
        )
    )
    .reset_index()
)

role_skill_summary = role_skill_summary.sort_values(
    by="Total_Skills",
    ascending=False
).reset_index(drop=True)

print("ROLE SKILL REQUIREMENT SUMMARY")
print("=" * 70)

display(
    role_skill_summary.head(20)
)
```

```python
# ============================================================
# HIGH-DEMAND SOFTWARE SKILLS
# ============================================================

high_demand_software = software_skills[
    software_skills["In Demand"].astype(str).str.upper() == "Y"
].copy()

high_demand_summary = (
    high_demand_software
    .groupby("Element Name")
    .agg(
        Occupation_Count=("O*NET-SOC Code", "nunique")
    )
    .reset_index()
    .sort_values(
        by="Occupation_Count",
        ascending=False
    )
    .reset_index(drop=True)
)

print("TOP HIGH-DEMAND SOFTWARE SKILLS")
print("=" * 70)

display(
    high_demand_summary.head(20)
)
```

```python
# ============================================================
# SAVE WORKFORCE INTELLIGENCE OUTPUTS
# ============================================================

engagement_output = (
    OUTPUT_PATH / "department_engagement_intelligence.csv"
)

role_output = (
    OUTPUT_PATH / "job_role_intelligence.csv"
)

occupation_output = (
    OUTPUT_PATH / "occupation_intelligence.csv"
)

skill_output = (
    OUTPUT_PATH / "role_skill_intelligence.csv"
)

summary_output = (
    OUTPUT_PATH / "role_skill_summary.csv"
)

software_output = (
    OUTPUT_PATH / "high_demand_software_skills.csv"
)


department_engagement.to_csv(
    engagement_output,
    index=False
)

role_intelligence.to_csv(
    role_output,
    index=False
)

occupation_intelligence.to_csv(
    occupation_output,
    index=False
)

role_skill_requirements.to_csv(
    skill_output,
    index=False
)

role_skill_summary.to_csv(
    summary_output,
    index=False
)

high_demand_summary.to_csv(
    software_output,
    index=False
)

print("✓ All Workforce Intelligence outputs saved.")

print("\nFiles:")
print(f"✓ {engagement_output.name}")
print(f"✓ {role_output.name}")
print(f"✓ {occupation_output.name}")
print(f"✓ {skill_output.name}")
print(f"✓ {summary_output.name}")
print(f"✓ {software_output.name}")
```
