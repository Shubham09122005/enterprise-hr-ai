```python
# ============================================================
# FEATURE ENGINEERING
# Enterprise HR AI Platform
# ============================================================

import pandas as pd
import numpy as np

from pathlib import Path

pd.set_option("display.max_columns", None)

print("Libraries imported successfully.")
```

```python
# ============================================================
# PROJECT PATH SETUP
# ============================================================

CURRENT_DIR = Path.cwd()

PROJECT_ROOT = CURRENT_DIR.parent

PROCESSED_DATA_PATH = PROJECT_ROOT / "data" / "processed"

print("Current Directory:")
print(CURRENT_DIR)

print("\nProject Root:")
print(PROJECT_ROOT)

print("\nProcessed Data Path:")
print(PROCESSED_DATA_PATH)

print("\nProcessed Data Exists:")
print(PROCESSED_DATA_PATH.exists())
```

```python
# ============================================================
# LOAD CLEANED DATASETS
# ============================================================

datasets = {
    "employee_attrition": PROCESSED_DATA_PATH / "employee_attrition_cleaned.csv",
    "performance_engagement": PROCESSED_DATA_PATH / "performance_engagement_cleaned.csv",
    "occupational_data": PROCESSED_DATA_PATH / "occupational_data_cleaned.csv",
    "essential_skills": PROCESSED_DATA_PATH / "essential_skills_cleaned.csv",
    "software_skills": PROCESSED_DATA_PATH / "software_skills_cleaned.csv"
}

cleaned_data = {}

for name, path in datasets.items():
    if path.exists():
        cleaned_data[name] = pd.read_csv(path)
        print(f"✓ Loaded {name}: {cleaned_data[name].shape}")
    else:
        print(f"✗ File not found: {path}")
```

```python
# ============================================================
# DATASET OVERVIEW
# ============================================================

for name, df in cleaned_data.items():
    
    print("\n" + "=" * 70)
    print(f"DATASET: {name.upper()}")
    print("=" * 70)
    
    print(f"Shape: {df.shape}")
    
    print("\nFirst 5 rows:")
    display(df.head())
```

```python
# ============================================================
# DATASET OVERVIEW
# ============================================================

dataset_overview = []

for name, df in cleaned_data.items():
    dataset_overview.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })

dataset_overview_df = pd.DataFrame(dataset_overview)

print("DATASET OVERVIEW")
print("=" * 70)

display(dataset_overview_df)
```

```python
# ============================================================
# CREATE WORKING COPIES FOR FEATURE ENGINEERING
# ============================================================

employee_features = cleaned_data["employee_attrition"].copy()

performance_features = cleaned_data["performance_engagement"].copy()

occupational_features = cleaned_data["occupational_data"].copy()

essential_skills_features = cleaned_data["essential_skills"].copy()

software_skills_features = cleaned_data["software_skills"].copy()

print("✓ Working copies created successfully")

print(f"\nEmployee Features Shape: {employee_features.shape}")
print(f"Performance Features Shape: {performance_features.shape}")
print(f"Occupational Features Shape: {occupational_features.shape}")
print(f"Essential Skills Features Shape: {essential_skills_features.shape}")
print(f"Software Skills Features Shape: {software_skills_features.shape}")
```

```python
# ============================================================
# FEATURE ENGINEERING: AGE GROUPS
# ============================================================

age_bins = [0, 25, 35, 45, 55, 100]

age_labels = [
    "18-25",
    "26-35",
    "36-45",
    "46-55",
    "56+"
]

employee_features["AgeGroup"] = pd.cut(
    employee_features["Age"],
    bins=age_bins,
    labels=age_labels,
    include_lowest=True
)

print("AGE GROUP FEATURE CREATED")

display(
    employee_features[
        ["Age", "AgeGroup"]
    ].head(10)
)

print("\nAge Group Distribution:")

display(
    employee_features["AgeGroup"]
    .value_counts()
    .sort_index()
    .to_frame(name="Employee Count")
)
```

```python
# ============================================================
# FEATURE ENGINEERING: INCOME CATEGORIES
# ============================================================

income_bins = [0, 3000, 6000, 10000, 15000, float("inf")]

income_labels = [
    "Low Income",
    "Lower-Middle Income",
    "Middle Income",
    "Upper-Middle Income",
    "High Income"
]

employee_features["IncomeCategory"] = pd.cut(
    employee_features["MonthlyIncome"],
    bins=income_bins,
    labels=income_labels,
    include_lowest=True
)

print("INCOME CATEGORY FEATURE CREATED")

display(
    employee_features[
        ["MonthlyIncome", "IncomeCategory"]
    ].head(10)
)

print("\nIncome Category Distribution:")

display(
    employee_features["IncomeCategory"]
    .value_counts()
    .sort_index()
    .to_frame(name="Employee Count")
)
```

```python
# ============================================================
# FEATURE ENGINEERING: EXPERIENCE GROUPS
# ============================================================

experience_bins = [-1, 2, 5, 10, 20, float("inf")]

experience_labels = [
    "0-2 Years",
    "3-5 Years",
    "6-10 Years",
    "11-20 Years",
    "20+ Years"
]

employee_features["ExperienceGroup"] = pd.cut(
    employee_features["TotalWorkingYears"],
    bins=experience_bins,
    labels=experience_labels
)

company_tenure_bins = [-1, 1, 3, 5, 10, float("inf")]

company_tenure_labels = [
    "0-1 Year",
    "2-3 Years",
    "4-5 Years",
    "6-10 Years",
    "10+ Years"
]

employee_features["CompanyTenureGroup"] = pd.cut(
    employee_features["YearsAtCompany"],
    bins=company_tenure_bins,
    labels=company_tenure_labels
)

print("EXPERIENCE FEATURES CREATED")

display(
    employee_features[
        [
            "TotalWorkingYears",
            "ExperienceGroup",
            "YearsAtCompany",
            "CompanyTenureGroup"
        ]
    ].head(10)
)
```

```python
# ============================================================
# FEATURE ENGINEERING: EMPLOYEE SATISFACTION SCORE
# ============================================================

satisfaction_columns = [
    "EnvironmentSatisfaction",
    "JobSatisfaction",
    "RelationshipSatisfaction",
    "WorkLifeBalance"
]

employee_features["OverallSatisfactionScore"] = (
    employee_features[satisfaction_columns]
    .mean(axis=1)
    .round(2)
)

def satisfaction_category(score):
    if score <= 2:
        return "Low Satisfaction"
    elif score <= 3:
        return "Medium Satisfaction"
    else:
        return "High Satisfaction"


employee_features["SatisfactionCategory"] = (
    employee_features["OverallSatisfactionScore"]
    .apply(satisfaction_category)
)

print("SATISFACTION FEATURES CREATED")

display(
    employee_features[
        satisfaction_columns +
        [
            "OverallSatisfactionScore",
            "SatisfactionCategory"
        ]
    ].head(10)
)
```

```python
# ============================================================
# FEATURE ENGINEERING: EMPLOYEE ATTRITION RISK FEATURES
# ============================================================

employee_features["OverTimeRisk"] = np.where(
    employee_features["OverTime"] == "Yes",
    "High Risk",
    "Lower Risk"
)

employee_features["LongCommuteRisk"] = np.where(
    employee_features["DistanceFromHome"] > 20,
    "High Risk",
    "Lower Risk"
)

employee_features["LowIncomeRisk"] = np.where(
    employee_features["MonthlyIncome"] < 3000,
    "High Risk",
    "Lower Risk"
)

employee_features["FrequentJobChangeRisk"] = np.where(
    employee_features["NumCompaniesWorked"] >= 4,
    "High Risk",
    "Lower Risk"
)

print("ATTRITION RISK FEATURES CREATED")

display(
    employee_features[
        [
            "OverTime",
            "OverTimeRisk",
            "DistanceFromHome",
            "LongCommuteRisk",
            "MonthlyIncome",
            "LowIncomeRisk",
            "NumCompaniesWorked",
            "FrequentJobChangeRisk"
        ]
    ].head(10)
)
```

```python
# ============================================================
# FEATURE ENGINEERING: COMPOSITE ATTRITION RISK SCORE
# ============================================================

employee_features["AttritionRiskScore"] = 0

# Overtime Risk
employee_features["AttritionRiskScore"] += np.where(
    employee_features["OverTime"] == "Yes",
    2,
    0
)

# Long Commute Risk
employee_features["AttritionRiskScore"] += np.where(
    employee_features["DistanceFromHome"] > 20,
    1,
    0
)

# Low Satisfaction Risk
employee_features["AttritionRiskScore"] += np.where(
    employee_features["OverallSatisfactionScore"] <= 2.5,
    2,
    0
)

# Low Income Risk
employee_features["AttritionRiskScore"] += np.where(
    employee_features["MonthlyIncome"] < 3000,
    1,
    0
)

# Frequent Job Change Risk
employee_features["AttritionRiskScore"] += np.where(
    employee_features["NumCompaniesWorked"] >= 4,
    1,
    0
)

def risk_category(score):
    if score <= 1:
        return "Low Risk"
    elif score <= 3:
        return "Medium Risk"
    else:
        return "High Risk"


employee_features["AttritionRiskCategory"] = (
    employee_features["AttritionRiskScore"]
    .apply(risk_category)
)

print("COMPOSITE ATTRITION RISK FEATURES CREATED")

display(
    employee_features[
        [
            "AttritionRiskScore",
            "AttritionRiskCategory"
        ]
    ].head(10)
)

print("\nRisk Category Distribution:")

display(
    employee_features["AttritionRiskCategory"]
    .value_counts()
    .to_frame(name="Employee Count")
)
```

```python
# ============================================================
# FEATURE ENGINEERING: PERFORMANCE & ENGAGEMENT FEATURES
# ============================================================

# Performance Category
performance_features["PerformanceCategory"] = pd.cut(
    performance_features["Performance Score"],
    bins=[0, 50, 70, 85, 100],
    labels=[
        "Low",
        "Average",
        "Good",
        "Excellent"
    ],
    include_lowest=True
)

# KPI Category
performance_features["KPICategory"] = pd.cut(
    performance_features["KPI Score"],
    bins=[0, 50, 70, 85, 100],
    labels=[
        "Low",
        "Average",
        "Good",
        "Excellent"
    ],
    include_lowest=True
)

# Attendance Category
performance_features["AttendanceCategory"] = pd.cut(
    performance_features["Attendance (%)"],
    bins=[0, 70, 85, 95, 100],
    labels=[
        "Low Attendance",
        "Moderate Attendance",
        "Good Attendance",
        "Excellent Attendance"
    ],
    include_lowest=True
)

# Task Completion Category
performance_features["TaskCompletionCategory"] = pd.cut(
    performance_features["Task Completion (%)"],
    bins=[0, 50, 70, 85, 100],
    labels=[
        "Low",
        "Average",
        "Good",
        "Excellent"
    ],
    include_lowest=True
)

print("PERFORMANCE AND ENGAGEMENT FEATURES CREATED")

display(
    performance_features[
        [
            "Performance Score",
            "PerformanceCategory",
            "KPI Score",
            "KPICategory",
            "Attendance (%)",
            "AttendanceCategory",
            "Task Completion (%)",
            "TaskCompletionCategory"
        ]
    ].head(10)
)
```

```python
# ============================================================
# FEATURE ENGINEERING: OVERALL PERFORMANCE INDEX
# ============================================================

performance_features["OverallPerformanceIndex"] = (
    (
        performance_features["Performance Score"] * 0.30
    ) +
    (
        performance_features["KPI Score"] * 0.25
    ) +
    (
        performance_features["Attendance (%)"] * 0.15
    ) +
    (
        performance_features["Peer Rating"] * 10 * 0.10
    ) +
    (
        performance_features["Task Completion (%)"] * 0.20
    )
).round(2)


def performance_index_category(score):
    if score < 60:
        return "Needs Improvement"
    elif score < 75:
        return "Average Performer"
    elif score < 90:
        return "High Performer"
    else:
        return "Top Performer"


performance_features["PerformanceIndexCategory"] = (
    performance_features["OverallPerformanceIndex"]
    .apply(performance_index_category)
)

print("OVERALL PERFORMANCE INDEX CREATED")

display(
    performance_features[
        [
            "Performance Score",
            "KPI Score",
            "Attendance (%)",
            "Peer Rating",
            "Task Completion (%)",
            "OverallPerformanceIndex",
            "PerformanceIndexCategory"
        ]
    ].head(10)
)

print("\nPerformance Index Category Distribution:")

display(
    performance_features["PerformanceIndexCategory"]
    .value_counts()
    .to_frame(name="Employee Count")
)

print("\nFINAL FEATURE ENGINEERING SHAPES")
print("=" * 60)

print(f"Employee Features Shape: {employee_features.shape}")
print(f"Performance Features Shape: {performance_features.shape}")
```

```python
# ============================================================
# FEATURE ENGINEERING: OCCUPATIONAL DATA
# ============================================================

occupational_features["TitleLength"] = (
    occupational_features["Title"]
    .astype(str)
    .str.len()
)

occupational_features["TitleWordCount"] = (
    occupational_features["Title"]
    .astype(str)
    .str.split()
    .str.len()
)

occupational_features["DescriptionLength"] = (
    occupational_features["Description"]
    .astype(str)
    .str.len()
)

occupational_features["DescriptionWordCount"] = (
    occupational_features["Description"]
    .astype(str)
    .str.split()
    .str.len()
)

print("OCCUPATIONAL FEATURES CREATED")

display(
    occupational_features[
        [
            "O*NET-SOC Code",
            "Title",
            "TitleLength",
            "TitleWordCount",
            "DescriptionLength",
            "DescriptionWordCount"
        ]
    ].head(10)
)
```

```python
# ============================================================
# OCCUPATIONAL DATA SUMMARY
# ============================================================

occupational_summary = pd.DataFrame({
    "Metric": [
        "Total Occupations",
        "Average Title Length",
        "Average Description Length",
        "Average Description Word Count"
    ],
    
    "Value": [
        len(occupational_features),
        round(occupational_features["TitleLength"].mean(), 2),
        round(occupational_features["DescriptionLength"].mean(), 2),
        round(occupational_features["DescriptionWordCount"].mean(), 2)
    ]
})

print("OCCUPATIONAL DATA SUMMARY")
print("=" * 60)

display(occupational_summary)
```

```python
# ============================================================
# FEATURE ENGINEERING: ESSENTIAL SKILLS PREPARATION
# ============================================================

essential_skills_features["Data Value"] = pd.to_numeric(
    essential_skills_features["Data Value"],
    errors="coerce"
)

essential_skills_features["SkillNameLength"] = (
    essential_skills_features["Element Name"]
    .astype(str)
    .str.len()
)

essential_skills_features["SkillNameWordCount"] = (
    essential_skills_features["Element Name"]
    .astype(str)
    .str.split()
    .str.len()
)

print("ESSENTIAL SKILLS PREPARATION COMPLETED")

print("\nData Value Statistics:")
display(
    essential_skills_features["Data Value"]
    .describe()
    .to_frame(name="Data Value")
)

display(
    essential_skills_features[
        [
            "O*NET-SOC Code",
            "Title",
            "Element Name",
            "Data Value",
            "SkillNameLength",
            "SkillNameWordCount"
        ]
    ].head(10)
)
```

```python
# ============================================================
# FEATURE ENGINEERING: SKILL IMPORTANCE CATEGORY
# ============================================================

data_value_min = essential_skills_features["Data Value"].min()
data_value_max = essential_skills_features["Data Value"].max()

print(f"Minimum Data Value: {data_value_min}")
print(f"Maximum Data Value: {data_value_max}")

essential_skills_features["SkillImportanceCategory"] = pd.qcut(
    essential_skills_features["Data Value"],
    q=4,
    labels=[
        "Low Importance",
        "Medium Importance",
        "High Importance",
        "Very High Importance"
    ],
    duplicates="drop"
)

print("\nSKILL IMPORTANCE CATEGORY CREATED")

display(
    essential_skills_features[
        [
            "Element Name",
            "Data Value",
            "SkillImportanceCategory"
        ]
    ].head(15)
)

print("\nSkill Importance Distribution:")

display(
    essential_skills_features["SkillImportanceCategory"]
    .value_counts()
    .sort_index()
    .to_frame(name="Skill Count")
)
```

```python
# ============================================================
# FEATURE ENGINEERING: OCCUPATION SKILL SUMMARY
# ============================================================

essential_skills_summary = (
    essential_skills_features
    .groupby(
        ["O*NET-SOC Code", "Title"]
    )
    .agg(
        TotalSkillRecords=("Element Name", "count"),
        UniqueSkills=("Element Name", "nunique"),
        AverageSkillValue=("Data Value", "mean"),
        MaximumSkillValue=("Data Value", "max"),
        MinimumSkillValue=("Data Value", "min")
    )
    .reset_index()
)

essential_skills_summary["AverageSkillValue"] = (
    essential_skills_summary["AverageSkillValue"]
    .round(2)
)

print("ESSENTIAL SKILLS OCCUPATION SUMMARY CREATED")

print(f"\nTotal Occupations in Summary: {len(essential_skills_summary)}")

display(
    essential_skills_summary.head(10)
)
```

```python
# ============================================================
# FEATURE ENGINEERING: SOFTWARE SKILLS PREPARATION
# ============================================================

software_skills_features["Hot Technology Normalized"] = (
    software_skills_features["Hot Technology"]
    .astype(str)
    .str.strip()
    .str.lower()
)

software_skills_features["In Demand Normalized"] = (
    software_skills_features["In Demand"]
    .astype(str)
    .str.strip()
    .str.lower()
)

print("SOFTWARE SKILLS PREPARATION COMPLETED")

print("\nHot Technology Values:")
print(
    software_skills_features["Hot Technology Normalized"]
    .value_counts()
    .head(10)
)

print("\nIn Demand Values:")
print(
    software_skills_features["In Demand Normalized"]
    .value_counts()
    .head(10)
)
```

```python
# ============================================================
# FEATURE ENGINEERING: SOFTWARE DEMAND FEATURES
# ============================================================

positive_values = [
    "yes",
    "y",
    "true",
    "1"
]

software_skills_features["IsHotTechnology"] = (
    software_skills_features["Hot Technology Normalized"]
    .isin(positive_values)
    .astype(int)
)

software_skills_features["IsInDemand"] = (
    software_skills_features["In Demand Normalized"]
    .isin(positive_values)
    .astype(int)
)

software_skills_features["TechnologyDemandScore"] = (
    software_skills_features["IsHotTechnology"] +
    software_skills_features["IsInDemand"]
)

def technology_demand_category(score):
    
    if score == 2:
        return "Hot & In Demand"
    
    elif score == 1:
        return "Moderate Demand"
    
    else:
        return "Standard"


software_skills_features["TechnologyDemandCategory"] = (
    software_skills_features["TechnologyDemandScore"]
    .apply(technology_demand_category)
)

print("SOFTWARE DEMAND FEATURES CREATED")

display(
    software_skills_features[
        [
            "Element Name",
            "Hot Technology",
            "In Demand",
            "IsHotTechnology",
            "IsInDemand",
            "TechnologyDemandScore",
            "TechnologyDemandCategory"
        ]
    ].head(15)
)
```

```python
# ============================================================
# FEATURE ENGINEERING: SOFTWARE SKILLS OCCUPATION SUMMARY
# ============================================================

software_skills_summary = (
    software_skills_features
    .groupby(
        ["O*NET-SOC Code", "Title"]
    )
    .agg(
        TotalSoftwareSkillRecords=("Element Name", "count"),
        UniqueSoftwareSkills=("Element Name", "nunique"),
        HotTechnologies=("IsHotTechnology", "sum"),
        InDemandTechnologies=("IsInDemand", "sum"),
        AverageTechnologyDemandScore=(
            "TechnologyDemandScore",
            "mean"
        )
    )
    .reset_index()
)

software_skills_summary[
    "AverageTechnologyDemandScore"
] = (
    software_skills_summary[
        "AverageTechnologyDemandScore"
    ].round(2)
)

print("SOFTWARE SKILLS OCCUPATION SUMMARY CREATED")

display(
    software_skills_summary.head(10)
)
```

```python
# ============================================================
# CREATE COMBINED OCCUPATION INTELLIGENCE DATASET
# ============================================================

occupation_intelligence = occupational_features.merge(
    
    essential_skills_summary[
        [
            "O*NET-SOC Code",
            "TotalSkillRecords",
            "UniqueSkills",
            "AverageSkillValue",
            "MaximumSkillValue",
            "MinimumSkillValue"
        ]
    ],
    
    on="O*NET-SOC Code",
    how="left"
)

occupation_intelligence = occupation_intelligence.merge(
    
    software_skills_summary[
        [
            "O*NET-SOC Code",
            "TotalSoftwareSkillRecords",
            "UniqueSoftwareSkills",
            "HotTechnologies",
            "InDemandTechnologies",
            "AverageTechnologyDemandScore"
        ]
    ],
    
    on="O*NET-SOC Code",
    how="left"
)

print("COMBINED OCCUPATION INTELLIGENCE DATASET CREATED")

print(f"\nShape: {occupation_intelligence.shape}")

display(
    occupation_intelligence.head(10)
)
```

```python
# ============================================================
# FINAL FEATURE ENGINEERING SUMMARY
# ============================================================

feature_engineering_summary = pd.DataFrame([
    
    {
        "Dataset": "Employee Attrition Features",
        "Rows": employee_features.shape[0],
        "Columns": employee_features.shape[1],
        "New Features": employee_features.shape[1] - 35
    },
    
    {
        "Dataset": "Performance Engagement Features",
        "Rows": performance_features.shape[0],
        "Columns": performance_features.shape[1],
        "New Features": performance_features.shape[1] - 13
    },
    
    {
        "Dataset": "Occupational Features",
        "Rows": occupational_features.shape[0],
        "Columns": occupational_features.shape[1],
        "New Features": occupational_features.shape[1] - 3
    },
    
    {
        "Dataset": "Essential Skills Features",
        "Rows": essential_skills_features.shape[0],
        "Columns": essential_skills_features.shape[1],
        "New Features": essential_skills_features.shape[1] - 15
    },
    
    {
        "Dataset": "Software Skills Features",
        "Rows": software_skills_features.shape[0],
        "Columns": software_skills_features.shape[1],
        "New Features": software_skills_features.shape[1] - 7
    },
    
    {
        "Dataset": "Occupation Intelligence",
        "Rows": occupation_intelligence.shape[0],
        "Columns": occupation_intelligence.shape[1],
        "New Features": "Combined Dataset"
    }
])

print("FINAL FEATURE ENGINEERING SUMMARY")
print("=" * 80)

display(feature_engineering_summary)
```

```python
# ============================================================
# SAVE FEATURE ENGINEERED DATASETS
# ============================================================

FEATURE_ENGINEERED_PATH = PROJECT_ROOT / "data" / "feature_engineered"

FEATURE_ENGINEERED_PATH.mkdir(
    parents=True,
    exist_ok=True
)

datasets_to_save = {
    "employee_features.csv": employee_features,
    "performance_features.csv": performance_features,
    "occupational_features.csv": occupational_features,
    "essential_skills_features.csv": essential_skills_features,
    "software_skills_features.csv": software_skills_features,
    "occupation_intelligence.csv": occupation_intelligence
}

for filename, dataframe in datasets_to_save.items():

    save_path = FEATURE_ENGINEERED_PATH / filename

    dataframe.to_csv(
        save_path,
        index=False
    )

    print(
        f"✓ Saved {filename} "
        f"{dataframe.shape}"
    )

print("\nFeature engineered datasets saved successfully.")
print(f"\nLocation:\n{FEATURE_ENGINEERED_PATH}")
```
