```python
from pathlib import Path
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 150)

print("Libraries imported successfully.")
```

```python
PROJECT_ROOT = Path.cwd().resolve().parent
DATA_DIR = PROJECT_ROOT / "data" / "raw"

print("Current working directory:")
print(Path.cwd())

print("\nProject root:")
print(PROJECT_ROOT)

print("\nRaw data directory:")
print(DATA_DIR)

print("\nDirectory exists:", DATA_DIR.exists())
```

```python
FILES = {
    "employee_attrition": "employee_attrition.csv",
    "performance_engagement": "hr_performance_engagement.csv",
    "occupational_data": "occupation_data.csv",
    "essential_skills": "essential_skills.csv",
    "software_skills": "software_skills.csv"
}

datasets = {}

for name, filename in FILES.items():
    file_path = DATA_DIR / filename
    
    if not file_path.exists():
        print(f"❌ File not found: {filename}")
    else:
        datasets[name] = pd.read_csv(file_path)
        print(f"✅ Loaded {filename} | Shape: {datasets[name].shape}")
```

```python
summary_rows = []

for name, df in datasets.items():
    
    summary_rows.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": int(df.isna().sum().sum()),
        "Duplicate Rows": int(df.duplicated().sum()),
        "Empty Columns": int(df.isna().all().sum())
    })

validation_summary = pd.DataFrame(summary_rows)

validation_summary
```

```python
attrition = datasets["employee_attrition"].copy()

print("Dataset Shape:")
print(attrition.shape)

print("\nColumn Names:")
print(attrition.columns.tolist())

print("\nFirst 5 Records:")
display(attrition.head())
```

```python
print("Data Types:\n")

display(
    attrition.dtypes.to_frame(name="Data Type")
)
```

```python
missing_values = attrition.isna().sum()

print("Missing Values by Column:\n")

display(
    missing_values[missing_values > 0].to_frame(
        name="Missing Values"
    )
)

print("\nTotal Missing Values:", attrition.isna().sum().sum())
```

```python
duplicate_rows = attrition.duplicated().sum()

print("Total Duplicate Rows:", duplicate_rows)

if duplicate_rows == 0:
    print("Status: ✅ PASS - No duplicate rows found")
else:
    print("Status: ⚠️ REVIEW - Duplicate rows detected")
```

```python
if "EmployeeNumber" in attrition.columns:
    
    duplicate_employee_ids = attrition["EmployeeNumber"].duplicated().sum()
    
    print("Total Employees:", len(attrition))
    print("Unique Employee Numbers:", attrition["EmployeeNumber"].nunique())
    print("Duplicate Employee Numbers:", duplicate_employee_ids)
    
    if duplicate_employee_ids == 0:
        print("\nStatus: ✅ PASS - EmployeeNumber is unique")
    else:
        print("\nStatus: ❌ FAIL - Duplicate EmployeeNumber values found")
        
else:
    print("⚠️ EmployeeNumber column not found")
```

```python
range_checks = {
    "Age": (15, 100),
    "MonthlyIncome": (0, None),
    "YearsAtCompany": (0, 100),
    "JobSatisfaction": (1, 4),
    "WorkLifeBalance": (1, 4)
}

range_results = []

for column, limits in range_checks.items():
    
    if column in attrition.columns:
        
        min_value, max_value = limits
        
        invalid_records = attrition[column] < min_value
        
        if max_value is not None:
            invalid_records = invalid_records | (
                attrition[column] > max_value
            )
        
        invalid_count = invalid_records.sum()
        
        range_results.append({
            "Column": column,
            "Expected Range": f"{min_value} to {max_value if max_value else '∞'}",
            "Invalid Records": invalid_count,
            "Status": "PASS" if invalid_count == 0 else "REVIEW"
        })

range_validation = pd.DataFrame(range_results)

range_validation
```

```python
categorical_checks = {
    "Attrition": {"Yes", "No"},
    "OverTime": {"Yes", "No"}
}

for column, expected_values in categorical_checks.items():
    
    if column in attrition.columns:
        
        actual_values = set(
            attrition[column]
            .dropna()
            .unique()
        )
        
        unexpected_values = actual_values - expected_values
        
        print(f"\nColumn: {column}")
        print("Actual Values:", actual_values)
        
        if len(unexpected_values) == 0:
            print("Status: ✅ PASS")
        else:
            print(
                "Status: ⚠️ REVIEW"
            )
            print(
                "Unexpected Values:",
                unexpected_values
            )
```

```python
print("Potential common columns between datasets:\n")

dataset_names = list(datasets.keys())

for i in range(len(dataset_names)):
    
    for j in range(i + 1, len(dataset_names)):
        
        dataset_a = dataset_names[i]
        dataset_b = dataset_names[j]
        
        columns_a = set(
            datasets[dataset_a].columns
        )
        
        columns_b = set(
            datasets[dataset_b].columns
        )
        
        common_columns = (
            columns_a.intersection(columns_b)
        )
        
        print(
            f"{dataset_a} ↔ {dataset_b}"
        )
        
        print(
            "Common Columns:",
            list(common_columns)
        )
        
        print("-" * 50)
```

```python
final_validation = []

for name, df in datasets.items():
    
    total_missing = int(
        df.isna().sum().sum()
    )
    
    duplicate_rows = int(
        df.duplicated().sum()
    )
    
    if total_missing == 0 and duplicate_rows == 0:
        status = "PASS"
    
    elif duplicate_rows > 0:
        status = "REVIEW"
    
    else:
        status = "REVIEW"
    
    final_validation.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": total_missing,
        "Duplicate Rows": duplicate_rows,
        "Validation Status": status
    })

final_validation_report = pd.DataFrame(
    final_validation
)

final_validation_report
```
